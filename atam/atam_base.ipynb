```python
import os
from pathlib import Path

# Ensure we are at repo root
while not Path("data").exists():
    os.chdir("..")

from atam.pipeline_base import run_base_pipeline

DATA_ROOT = Path("data")
OUTPUT_ROOT = Path("outputs")
PROMPT_PATH = Path("atam") / "prompts" / "base_scenario_mapping.md"

# Discover domains automatically
domains = [
    d for d in DATA_ROOT.iterdir()
    if d.is_dir()
    and d.name != "prompts"
    and (d / "scenarios.json").exists()
    and (d / "architectures").exists()
]

print(f"Discovered domains: {[d.name for d in domains]}")

run_dirs = {}

for domain_dir in domains:
    print(f"\n=== Running pipeline for domain: {domain_dir.name} ===")

    run_dir = run_base_pipeline(
        domain_dir=domain_dir,
        out_root=OUTPUT_ROOT,
        prompt_path=PROMPT_PATH,
        ollama_host="localhost:11434",
        ollama_model="llama3.2:3b",  # standard model, no custom creation
        max_retries=2,
    )

    run_dirs[domain_dir.name] = run_dir
    print(f"Completed domain '{domain_dir.name}', output in: {run_dir}")

print("\nAll pipeline runs completed.")
```
